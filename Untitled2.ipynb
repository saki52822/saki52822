```python
{
 "cells": [
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "# 104 大型科技業職缺推薦 (Colab 版)\n",
    "\n",
    "**使用方式**\n",
    "1. 左側鑰匙圖示 🔑「密鑰」新增 `GMAIL_USER`、`GMAIL_APP_PASSWORD`（可選 `MAIL_TO`），並開啟「筆記本存取權」。\n",
    "2. 依序執行 Cell 1 → 2 → 3。\n",
    "3. 最後一格先用 `DRY_RUN = True` 測試，確認有結果再改成 `False` 寄信。\n"
   ]
  },
  {
   "cell_type": "code",
   "metadata": {},
   "execution_count": none,
   "outputs": [],
   "source": [
    "!pip install -q requests"
   ]
  },
  {
   "cell_type": "code",
   "metadata": {},
   "execution_count": null,
   "outputs": [],
   "source": [
    "import os\n",
    "from google.colab import drive, userdata\n",
    "\n",
    "# 掛載雲端硬碟，用來記錄「已推薦過」的職缺（避免重複）\n",
    "drive.mount('/content/drive')\n",
    "os.makedirs('/content/drive/MyDrive/job_digest', exist_ok=True)\n",
    "\n",
    "# 從 Colab 密鑰讀取信箱資訊（不要把密碼直接寫在程式裡）\n",
    "for k in [\"GMAIL_USER\", \"GMAIL_APP_PASSWORD\", \"MAIL_TO\"]:\n",
    "    try:\n",
    "        os.environ[k] = userdata.get(k)\n",
    "    except Exception:\n",
    "        print(f\"（未設定 {k}）\")"
   ]
  },
  {
   "cell_type": "code",
   "metadata": {},
   "execution_count": null,
   "outputs": [],
   "source": [
    "#!/usr/bin/env python3\n",
    "\"\"\"\n",
    "104 大型科技業職缺推薦 → 寄到你的 Gmail\n",
    "用法:\n",
    "  python job_digest.py --dry-run     # 只在終端機顯示，不寄信（先用這個測試）\n",
    "  python job_digest.py               # 寄信\n",
    "\n",
    "需要的環境變數（寄信時）:\n",
    "  GMAIL_USER          你的 Gmail 帳號，例如 ruby52824@gmail.com\n",
    "  GMAIL_APP_PASSWORD  Gmail「應用程式密碼」(16 碼，不是登入密碼)\n",
    "  MAIL_TO             收件信箱（可省略，預設寄給自己）\n",
    "\n",
    "只需要 requests:  pip install requests\n",
    "\"\"\"\n",
    "import argparse\n",
    "import json\n",
    "import os\n",
    "import re\n",
    "import smtplib\n",
    "import ssl\n",
    "import time\n",
    "from datetime import datetime\n",
    "from email.mime.text import MIMEText\n",
    "from pathlib import Path\n",
    "\n",
    "import requests\n",
    "\n",
    "# ====================== 你可以自己改的設定 ======================\n",
    "\n",
    "KEYWORDS = [\n",
    "    \"助理專案經理\", \"軟體專案管理師\", \"專案管理師\", \"專案助理\", \"PMO\",\n",
    "    \"Junior PM\", \"Software PM\", \"Project Manager\",\n",
    "    \"商業數據分析\", \"營運數據分析\", \"數據分析師\", \"Business Data Analyst\", \"Data Analyst\",\n",
    "    \"助理產品經理\", \"產品經理\", \"Associate Product Manager\", \"Product Manager\",\n",
    "]\n",
    "\n",
    "# 台北市 / 新北市 (104 地區碼)\n",
    "AREAS = \"6001001000,6001002000\"\n",
    "\n",
    "# 優先區域：地址含這些字會標「⭐科學園區」\n",
    "PRIORITY_DISTRICTS = [\"南港\", \"內湖\", \"汐止\"]\n",
    "\n",
    "# 只保留這些區（空清單 = 不限，台北/新北全部）\n",
    "ONLY_DISTRICTS = []  # 例如 [\"南港\", \"內湖\", \"汐止\"]\n",
    "\n",
    "# 月薪下限（面議的職缺會保留並標示「面議」，請自行確認）\n",
    "MIN_SALARY = 50000\n",
    "\n",
    "# 大型科技公司白名單（公司名稱「包含」下列任一字串就算）\n",
    "# 104 搜尋結果沒有「公司規模」欄位，所以用白名單判斷「大型、知名」。\n",
    "# 想追加公司就直接加在這裡。\n",
    "BIG_TECH = [\n",
    "    # 半導體 / IC\n",
    "    \"台灣積體電路\", \"聯發科\", \"聯詠\", \"瑞昱\", \"聯電\", \"日月光\", \"群聯\", \"創意電子\",\n",
    "    \"世界先進\", \"力積電\", \"矽品\", \"穩懋\", \"義隆\", \"原相\", \"譜瑞\", \"祥碩\", \"旺宏\",\n",
    "    # 硬體 / 品牌 / 電子\n",
    "    \"華碩\", \"宏碁\", \"微星\", \"技嘉\", \"廣達\", \"緯創\", \"緯穎\", \"仁寶\", \"英業達\",\n",
    "    \"和碩\", \"鴻海\", \"富士康\", \"研華\", \"台達電\", \"光寶\", \"友達\", \"群創\", \"神達\", \"神基\",\n",
    "    # 軟體 / 網路 / 雲端 / 資安\n",
    "    \"趨勢科技\", \"精誠資訊\", \"叡揚\", \"關貿網路\", \"凌網\", \"數位無限\", \"Gogolook\", \"KKBOX\",\n",
    "    \"LINE\", \"蝦皮\", \"Shopee\", \"新加坡商\", \"酷澎\", \"Coupang\", \"foodpanda\", \"Uber\",\n",
    "    \"momo\", \"富邦媒體\", \"PChome\", \"網家\", \"91APP\", \"資策會\", \"鼎新\", \"凌群\", \"中華電信\",\n",
    "    \"遠傳\", \"台灣大哥大\", \"台灣之星\",\n",
    "    # 外商科技\n",
    "    \"Google\", \"谷歌\", \"Microsoft\", \"微軟\", \"Amazon\", \"亞馬遜\", \"AWS\", \"Apple\", \"蘋果\",\n",
    "    \"Meta\", \"Intel\", \"英特爾\", \"NVIDIA\", \"輝達\", \"Qualcomm\", \"高通\", \"AMD\", \"Cisco\",\n",
    "    \"思科\", \"IBM\", \"Oracle\", \"甲骨文\", \"SAP\", \"Synopsys\", \"新思\", \"Cadence\", \"益華\",\n",
    "    \"Micron\", \"美光\", \"Dell\", \"戴爾\", \"HP\", \"惠普\", \"Salesforce\", \"Adobe\", \"Garmin\", \"佳明\",\n",
    "    \"Texas Instruments\", \"德州儀器\", \"Marvell\", \"邁威爾\", \"Broadcom\", \"博通\",\n",
    "    \"Booking\", \"Agoda\", \"Trend Micro\", \"VMware\", \"ServiceNow\", \"Splunk\", \"Yahoo\", \"雅虎\",\n",
    "]\n",
    "\n",
    "# 標題含這些字 → 通常要求資深，直接排除（你是 1~2 年資歷）\n",
    "EXCLUDE_TITLE = r\"資深|Senior|Sr\\.|主任|協理|副理|總監|Director|VP|Head|Principal|Staff|Lead|主管|課長\"\n",
    "\n",
    "# 每個關鍵字抓幾頁（每頁約 20 筆）\n",
    "PAGES_PER_KEYWORD = 2\n",
    "\n",
    "# ================================================================\n",
    "\n",
    "SEEN_FILE = Path(\"/content/drive/MyDrive/job_digest/seen_jobs.json\")  # 存在你的 Google 雲端硬碟\n",
    "API = \"https://www.104.com.tw/jobs/search/api/jobs\"\n",
    "HEADERS = {\n",
    "    \"User-Agent\": \"Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 \"\n",
    "                  \"(KHTML, like Gecko) Chrome/124.0 Safari/537.36\",\n",
    "    \"Referer\": \"https://www.104.com.tw/jobs/search/\",\n",
    "    \"Accept\": \"application/json, text/plain, */*\",\n",
    "}\n",
    "\n",
    "\n",
    "def fetch(keyword, page):\n",
    "    params = {\n",
    "        \"keyword\": keyword,\n",
    "        \"area\": AREAS,\n",
    "        \"order\": 15,       # 依日期排序；若結果怪怪的可改成 1（符合度）\n",
    "        \"asc\": 0,\n",
    "        \"page\": page,\n",
    "        \"mode\": \"s\",\n",
    "        \"jobsource\": \"index_s\",\n",
    "        \"langFlag\": 0,\n",
    "    }\n",
    "    r = requests.get(API, params=params, headers=HEADERS, timeout=20)\n",
    "    r.raise_for_status()\n",
    "    return r.json().get(\"data\", {}).get(\"list\", [])\n",
    "\n",
    "\n",
    "def is_big_tech(company):\n",
    "    c = company.lower()\n",
    "    return any(name.lower() in c for name in BIG_TECH)\n",
    "\n",
    "\n",
    "def parse_job(j):\n",
    "    link = (j.get(\"link\") or {}).get(\"job\", \"\")\n",
    "    if link.startswith(\"//\"):\n",
    "        link = \"https:\" + link\n",
    "    low = int(j.get(\"salaryLow\") or 0)\n",
    "    high = int(j.get(\"salaryHigh\") or 0)\n",
    "    return {\n",
    "        \"id\": link.rstrip(\"/\").split(\"/\")[-1].split(\"?\")[0] or j.get(\"jobNo\", \"\"),\n",
    "        \"title\": j.get(\"jobName\", \"\"),\n",
    "        \"company\": j.get(\"custName\", \"\"),\n",
    "        \"addr\": (j.get(\"jobAddrNoDesc\", \"\") + j.get(\"jobAddress\", \"\")),\n",
    "        \"salary_low\": low,\n",
    "        \"salary_high\": high,\n",
    "        \"salary_desc\": j.get(\"salaryDesc\", \"\"),\n",
    "        \"date\": j.get(\"appearDate\", \"\"),\n",
    "        \"link\": link,\n",
    "    }\n",
    "\n",
    "\n",
    "def passes(job):\n",
    "    if not job[\"id\"] or not is_big_tech(job[\"company\"]):\n",
    "        return False\n",
    "    if re.search(EXCLUDE_TITLE, job[\"title\"], re.I):\n",
    "        return False\n",
    "    if ONLY_DISTRICTS and not any(d in job[\"addr\"] for d in ONLY_DISTRICTS):\n",
    "        return False\n",
    "    # 薪資：有明確數字時，上限要 >= MIN_SALARY；面議(0) 則保留\n",
    "    if job[\"salary_high\"] and job[\"salary_high\"] < MIN_SALARY:\n",
    "        return False\n",
    "    return True\n",
    "\n",
    "\n",
    "def collect():\n",
    "    found = {}\n",
    "    for kw in KEYWORDS:\n",
    "        for page in range(1, PAGES_PER_KEYWORD + 1):\n",
    "            try:\n",
    "                jobs = fetch(kw, page)\n",
    "            except Exception as e:  # noqa\n",
    "                print(f\"[警告] 搜尋「{kw}」第 {page} 頁失敗: {e}\")\n",
    "                break\n",
    "            if not jobs:\n",
    "                break\n",
    "            for j in jobs:\n",
    "                job = parse_job(j)\n",
    "                if job[\"id\"] and job[\"id\"] not in found and passes(job):\n",
    "                    found[job[\"id\"]] = job\n",
    "            time.sleep(1.2)  # 禮貌性延遲，避免被擋\n",
    "    return list(found.values())\n",
    "\n",
    "\n",
    "def load_seen():\n",
    "    if SEEN_FILE.exists():\n",
    "        return set(json.loads(SEEN_FILE.read_text(encoding=\"utf-8\")))\n",
    "    return set()\n",
    "\n",
    "\n",
    "def save_seen(seen):\n",
    "    SEEN_FILE.write_text(json.dumps(sorted(seen), ensure_ascii=False), encoding=\"utf-8\")\n",
    "\n",
    "\n",
    "def sort_key(job):\n",
    "    priority = any(d in job[\"addr\"] for d in PRIORITY_DISTRICTS)\n",
    "    return (not priority, job[\"company\"])\n",
    "\n",
    "\n",
    "def build_html(jobs):\n",
    "    rows = []\n",
    "    for j in sorted(jobs, key=sort_key):\n",
    "        star = \"⭐科學園區 \" if any(d in j[\"addr\"] for d in PRIORITY_DISTRICTS) else \"\"\n",
    "        sal = j[\"salary_desc\"] or (\"面議\" if not j[\"salary_high\"] else\n",
    "                                   f\"{j['salary_low']:,}~{j['salary_high']:,}\")\n",
    "        rows.append(\n",
    "            f\"<tr><td>{star}<b>{j['company']}</b></td>\"\n",
    "            f\"<td><a href='{j['link']}'>{j['title']}</a></td>\"\n",
    "            f\"<td>{j['addr'][:14]}</td><td>{sal}</td><td>{j['date']}</td></tr>\"\n",
    "        )\n",
    "    return (\n",
    "        f\"<h3>今日 104 新職缺（{datetime.now():%Y-%m-%d}）共 {len(jobs)} 筆</h3>\"\n",
    "        \"<table border='1' cellpadding='6' cellspacing='0' style='border-collapse:collapse'>\"\n",
    "        \"<tr><th>公司</th><th>職稱</th><th>地點</th><th>薪資</th><th>刊登日</th></tr>\"\n",
    "        + \"\".join(rows) + \"</table>\"\n",
    "        \"<p style='color:#888'>提醒：「面議」請自行確認是否達 5 萬；公司是否家族企業、\"\n",
    "        \"主管風格請另外查證。</p>\"\n",
    "    )\n",
    "\n",
    "\n",
    "def send_mail(html, count):\n",
    "    user = os.environ[\"GMAIL_USER\"]\n",
    "    pwd = os.environ[\"GMAIL_APP_PASSWORD\"]\n",
    "    to = os.environ.get(\"MAIL_TO\", user)\n",
    "    msg = MIMEText(html, \"html\", \"utf-8\")\n",
    "    msg[\"Subject\"] = f\"104 大型科技業新職缺 {count} 筆 ({datetime.now():%m/%d})\"\n",
    "    msg[\"From\"] = user\n",
    "    msg[\"To\"] = to\n",
    "    with smtplib.SMTP_SSL(\"smtp.gmail.com\", 465, context=ssl.create_default_context()) as s:\n",
    "        s.login(user, pwd)\n",
    "        s.send_message(msg)\n",
    "\n",
    "\n",
    "def main():\n",
    "    ap = argparse.ArgumentParser()\n",
    "    ap.add_argument(\"--dry-run\", action=\"store_true\", help=\"只顯示，不寄信、不記錄已讀\")\n",
    "    args = ap.parse_args([\"--dry-run\"] if DRY_RUN else [])\n",
    "\n",
    "    jobs = collect()\n",
    "    seen = load_seen()\n",
    "    new_jobs = [j for j in jobs if j[\"id\"] not in seen]\n",
    "    print(f\"符合條件 {len(jobs)} 筆，其中新職缺 {len(new_jobs)} 筆\")\n",
    "\n",
    "    if args.dry_run:\n",
    "        for j in sorted(new_jobs, key=sort_key):\n",
    "            print(f\"- {j['company']} | {j['title']} | {j['addr'][:12]} | \"\n",
    "                  f\"{j['salary_desc'] or '面議'} | {j['link']}\")\n",
    "        return\n",
    "\n",
    "    if not new_jobs:\n",
    "        print(\"今天沒有新職缺，不寄信。\")\n",
    "        return\n",
    "\n",
    "    send_mail(build_html(new_jobs), len(new_jobs))\n",
    "    seen.update(j[\"id\"] for j in new_jobs)\n",
    "    save_seen(seen)\n",
    "    print(\"已寄出。\")\n",
    "\n",
    "\n"
   ]
  },
  {
   "cell_type": "code",
   "metadata": {},
   "execution_count": null,
   "outputs": [],
   "source": [
    "DRY_RUN = True   # 測試用：只顯示不寄信。確認沒問題後改成 False\n",
    "main()"
   ]
  }
 ],
 "metadata": {
  "colab": {
   "provenance": []
  },
  "kernelspec": {
   "name": "python3",
   "display_name": "Python 3"
  },
  "language_info": {
   "name": "python"
  }
 },
 "nbformat": 4,
 "nbformat_minor": 0
}
```
